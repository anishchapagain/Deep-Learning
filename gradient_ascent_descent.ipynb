# Gradient

A gradient tells how a function's output changes as its inputs change.
- For a single-variable function, the gradient is the derivative (slope): 
    -   positive means the function increases as the input increases, 
    -   negative means it decreases.
- For multivariable functions, the gradient is a vector of partial derivatives. It points in the direction of
    steepest increase and its magnitude indicates the rate of increase in that direction.

# Gradient Ascent vs. Gradient Descent

This notebook explores two fundamental optimization algorithms, Gradient Descent and Gradient Ascent. These are iterative methods used to find the minimum or maximum of a function, respectively. They are the bedrock of how we train most machine learning models.

## 1. Gradient Descent

**Goal:** To find the **minimum** of a function.

**Concept:** Imagine you are standing on a mountain in a thick fog and you want to get to the lowest point in the valley. The most straightforward strategy is to look at the ground beneath your feet and take a step in the steepest downhill direction. You repeat this process until you can no longer step downwards, hopefully arriving at the bottom.

Gradient Descent does exactly this. It calculates the gradient (the direction of steepest *ascent*) and takes a step in the **opposite** direction to move towards a minimum.

**Mathematical Formula:**

The update rule for Gradient Descent is:

$\theta_{\text{new}} = \theta_{\text{old}} - \alpha \nabla f(\theta_{\text{old}})$

Where:
- $ \theta $: A parameter (or set of parameters) we are trying to optimize.
- $ \alpha $: The **learning rate**, a small positive number that controls the size of our step.
- $ \nabla f(\theta) $: The gradient of the function $f$ with respect to the parameters $ \theta $.
- The **minus sign** is crucial, as it ensures we move *against* the gradient, i.e., downhill.

**Applicability & Practical Use:**

- **The primary use case is minimizing a loss or cost function** during the training of a machine learning model.
- By minimizing the loss (the error between the model's predictions and the actual values), we are making the model more accurate.
- Used in Linear Regression, Logistic Regression, Neural Networks, and many other algorithms.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define a convex function (a parabola opening upwards) to minimize
def f(x):
    return x**2

# Define the gradient of the function
def gradient(x):
    return 2*x

# Gradient descent parameters
learning_rate = 0.1
num_iterations = 20

# Initialize a starting point
x = 4

# Store values for plotting
x_values = [x]
f_values = [f(x)]

# Perform gradient descent
for _ in range(num_iterations):
    x = x - learning_rate * gradient(x) # Note the minus '-' sign
    x_values.append(x)
    f_values.append(f(x))

# Plotting the results
x_range = np.linspace(-5, 5, 100)
plt.plot(x_range, f(x_range), label='f(x) = x^2')
plt.scatter(x_values, f_values, color='red', label='Gradient Descent Steps')
plt.title('Gradient Descent Example')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid()
plt.show()
```

## 2. Gradient Ascent

**Goal:** To find the **maximum** of a function.

**Concept:** Using the same mountain analogy, now your goal is to get to the highest peak. The strategy is to look for the steepest uphill direction and take a step. You repeat this process until you reach a peak.

Gradient Ascent calculates the gradient and takes a step in the **same** direction as the gradient to move towards a maximum.

**Mathematical Formula:**

The update rule for Gradient Ascent is:

$\theta_{\text{new}} = \theta_{\text{old}} + \alpha \nabla f(\theta_{\text{old}})$

Notice the only difference is the **plus sign**. We are now moving *along* the gradient direction.

**Applicability & Practical Use:**

- **Maximizing a likelihood function.** In statistics, Maximum Likelihood Estimation (MLE) is a common method for fitting a model to data, and it can be solved with gradient ascent.
- **Reinforcement Learning.** In policy gradient methods, an agent tries to maximize its expected future rewards by adjusting its policy using gradient ascent.
- **Generating Adversarial Examples.** To fool a neural network, one can slightly perturb an input image in a way that maximizes the model's loss, making it misclassify the image. This is done using gradient ascent on the loss function.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define a concave function (a parabola opening downwards) to maximize
def f(x):
    return -x**2 + 4*x

# Define the gradient of the function
def gradient(x):
    return -2*x + 4

# Gradient ascent parameters
learning_rate = 0.1
num_iterations = 20

# Initialize starting point
x = 0

# Store the values for plotting
x_values = [x]
f_values = [f(x)]

# Perform gradient ascent
for _ in range(num_iterations):
    x = x + learning_rate * gradient(x) # Note the plus '+' sign
    x_values.append(x)
    f_values.append(f(x))

# Plotting the results
x_range = np.linspace(-1, 5, 100)
plt.plot(x_range, f(x_range), label='f(x) = -x^2 + 4x')
plt.scatter(x_values, f_values, color='red', label='Gradient Ascent Steps')
plt.title('Gradient Ascent Example')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid()
plt.show()
```

## 3. Summary and Relationship

| | Gradient Descent | Gradient Ascent |
|---:|:---|:---|
| **Goal** | Minimize a function | Maximize a function |
| **Update Rule** | $ \theta - \alpha \nabla f(\theta) $ | $ \theta + \alpha \nabla f(\theta) $ |
| **Analogy** | Walking downhill into a valley | Climbing uphill to a peak |
| **Primary Use** | Minimizing loss in ML models | Maximizing likelihood or reward |

### The Core Relationship

The most important concept to understand is that these two algorithms are two sides of the same coin.

> **Maximizing a function `f(x)` is mathematically equivalent to minimizing its negative, `-f(x)`.**

This means you can always frame a maximization problem as a minimization problem. In fact, in the machine learning community, it is standard practice to almost always talk about **minimizing a loss function**. If you need to maximize something (like a likelihood), you simply define your loss as the *negative likelihood* and then use gradient descent.
