# Simple Neural Network

Simple neural network with the following structure:

*   **Input Layer:** 1 neuron
*   **Hidden Layer:** 1 neuron
*   **Output Layer:** 1 neuron

x  →  [Hidden neuron]  →  [Output neuron]

### Regression vs. Classification Output Layer

The choice of the output layer activation function depends on the task:

*   **Binary Classification:** Uses an activation like Sigmoid to output a value between 0 and 1, representing the probability of belonging to a specific class.
*   **Regression:** Typically uses a linear activation (or no activation function after the linear layer) to output a continuous value.

In the provided `SimpleNN` example, the `Sigmoid` in the output layer makes it suitable for a binary classification-like task.

To adapt it for regression, you can remove or change the `Sigmoid` activation.

## Equations and Calculations

Here's the computations within neural network:

**1. Calculation in the Hidden Layer:**

a. The **input** to the hidden neuron is the **weighted sum of the input neuron's value and a bias term**.

b. This **sum** is then **passed through an activation function** (commonly the sigmoid or ReLU function).

Let:
*   $x$ : the input value
*   $w_{ih}$ : the weight connecting the input neuron to the hidden neuron
*   $b_h$ : the bias for the hidden neuron
*   $z_h$ : the weighted sum before activation in the hidden layer
*   $a_h$ :the output of the hidden neuron after activation

The equations made are:

1. $z_h = x \cdot w_{ih} + b_h$

2. $a_h = \sigma(z_h)$  (using the sigmoid activation function $\sigma(z) = \frac{1}{1 + e^{-z}}$)

or

2. $a_h = \text{ReLU}(z_h)$ (using the ReLU activation function $\text{ReLU}(z) = \max(0, z)$)

<hr />

**2. Calculation in the Output Layer:**

The input to the output neuron is the weighted sum of the hidden neuron's output and a bias term. This sum is then passed through an activation function (depending on the task, e.g., sigmoid for binary classification, linear for regression).

Let:
*   $w_{ho}$ : the weight connecting the hidden neuron to the output neuron
*   $b_o$ : the bias for the output neuron
*   $z_o$ : the weighted sum before activation in the output layer
*   $a_o$ : the final output of the network

**Note**:
$a_h$: The output from hidden layer. It is now input to the current layer.
* ***Activation from one layer is carried as input to the next layer***.

The equations are:

1. $z_o = a_h \cdot w_{ho} + b_o$

2. $a_o = \sigma(z_o)$ (for binary classification)

or

2. $a_o = z_o$ (for regression)

```python
import torch
import torch.nn as nn # import
```

```python
# Define the simple neural network
class SimpleNN(nn.Module):
    def __init__(self):
        """
        Define the layers of the neural network.
        """
        super(SimpleNN, self).__init__()

        # Define the hidden layer: 1 input feature, 1 output feature
        self.hidden = nn.Linear(1, 1)

        # Define the output layer: 1 input feature (from hidden), 1 output feature
        self.output = nn.Linear(1, 1)

        # Using ReLU activation for the hidden layer
        self.relu = nn.ReLU()

        # Using Sigmoid activation for the output layer (for example, binary classification)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        """
        Define the forward pass of the neural network.
        """
        # Pass through the hidden layer and apply ReLU activation
        hidden_output = self.relu(self.hidden(x))

        # Pass through the output layer and apply Sigmoid activation
        final_output = self.sigmoid(self.output(hidden_output))

        return final_output
```

```python
# Create an instance of the Simple Neural network
model = SimpleNN()

# Print the model structure
print(model)
```

```python
# Example input (a single value)
input_data = torch.tensor([0.5], dtype=torch.float32)

# Pass the input through the model
output = model(input_data)

# Print the output
print(f"Input: {input_data.item()}")
print(f"Output: {output.item()}")
```

```python
# Access the weights and biases
print("Weights and Biases:")

for name, param in model.named_parameters():
    if param.requires_grad:
        print(f"{name}: {param.data}")
```
