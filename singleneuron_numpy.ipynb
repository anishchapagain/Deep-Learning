```python
import numpy as np
```

```python
x = np.array([0.6,0.4]) # two inputs
w = np.array([0.4,-0.2]) # two weights
b = 0.5 # bias
```

```python
z = np.dot(x,w)+b # Sum of input.weight & bias
```

```python
f"Output: No Activation {z}"
```

```python
# Activation Function

y_relu = np.maximum(0,z)
y_sigmoid = 1/(1+np.exp(-z))
y_tanh = np.tanh(z)
```

```python
print(f"Relu: Weighted sum: {z} -- Output after Activation function: {y_relu}")
print(f"Sigmoid: Weighted sum: {z} -- Output after Activation function: {y_sigmoid}")
print(f"Tanh: Weighted sum: {z} -- Output after Activation function: {y_tanh}")
```

```python
z1 = np.linspace(-8,8,100)
```

```python
z1
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
plt.plot(z1, 1/(1+np.exp(-z1)), label='Sigmoid', color='blue')
plt.plot(z1, np.maximum(0,z1), label='ReLU', color='green')
plt.plot(z1, np.tanh(z1), label='Tanh', color='red')

plt.title("Activation Functions (using linspace)")
plt.xlabel("z1 - Linear space")
plt.ylabel("Activation output")
plt.legend()
plt.grid(True)
plt.show()
```
