# Single Neuron with NumPy

## 1. Import NumPy
Import the NumPy library for numerical operations.

```python
import numpy as np
```

## 2. Define Inputs, Weights, and Bias
Define the input features, weights, and bias for the neuron.

```python
x = np.array([0.6,0.4]) # two inputs
w = np.array([0.4,-0.2]) # two weights
b = 0.5 # bias
```

## 3. Calculate the Weighted Sum
Calculate the weighted sum of the inputs and add the bias. This is the linear operation of the neuron.

```python
z = np.dot(x,w)+b # Sum of input.weight & bias
```

## 4. Output Before Activation
Display the output of the linear operation before applying any activation function.

```python
f'Output: No Activation {z}'
```

## 5. Apply Activation Functions
Apply different activation functions (ReLU, Sigmoid, Tanh) to the linear output.

```python
# Activation Function

y_relu = np.maximum(0,z)
y_sigmoid = 1/(1+np.exp(-z))
y_tanh = np.tanh(z)
```

## 6. Outputs After Activation
Display the outputs of the neuron after applying the activation functions.

```python
print(f'Relu: Weighted sum: {z} -- Output after Activation function: {y_relu}')
print(f'Sigmoid: Weighted sum: {z} -- Output after Activation function: {y_sigmoid}')
print(f'Tanh: Weighted sum: {z} -- Output after Activation function: {y_tanh}')
```

## 7. Visualize Activation Functions
Plot the activation functions to understand their behavior.

```python
import matplotlib.pyplot as plt
z_range = np.linspace(-5, 5, 100)
y_relu_range = np.maximum(0, z_range)
y_sigmoid_range = 1 / (1 + np.exp(-z_range))
y_tanh_range = np.tanh(z_range)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.plot(z_range, y_relu_range)
plt.title('ReLU')
plt.grid(True)

plt.subplot(1, 3, 2)
plt.plot(z_range, y_sigmoid_range)
plt.title('Sigmoid')
plt.grid(True)

plt.subplot(1, 3, 3)
plt.plot(z_range, y_tanh_range)
plt.title('Tanh')
plt.grid(True)

plt.show()
```

## Conclusion
This notebook demonstrated the basic operations of a single neuron using NumPy, including the linear combination of inputs and the application of common activation functions.
