# XOR Problem

The XOR problem is a classic example in deep learning which demonstrates that a single-layer perceptron cannot solve nonlinear problems like XOR. To solve XOR, a multi-layer perceptron (MLP) with at least one hidden layer and nonlinear activation functions is needed.

**Problem Overview**:
- XOR (exclusive OR) outputs 1 when the inputs differ, and 0 otherwise.

- It is not linearly separable, meaning a single-layer perceptron cannot learn XOR.

- Neural networks overcome this by transforming input space using hidden layers and nonlinear activation functions to create linearly separable features.
