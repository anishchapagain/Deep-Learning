# Feature Extraction Using MobileNet

```python
import torch
import torch.nn as nn
import torchvision.models as models
```

# Load a pretrained MobileNetV2 model

```python
model = models.mobilenet_v2(pretrained=True)
model.eval()
```

```python
# Freeze all layers parameters
for param in model.parameters():
    param.requires_grad = False
```

# MobileNetV2 has a classifier layer at the end, which we don't need for feature extraction.
# Assume we only want the features before the classifier layer.

# Remove the classifier layer

```python
num_ftrs = model.classifier[1].in_features
model.classifier[1] = nn.Linear(num_ftrs, 10) 
# Assume 10 classes
```

```python
model.classifier
```

```python
import torch.optim as optim
```

```python
# The optimizer is not needed for feature extraction
# we only need the model for feature extraction
optimizer = optim.Adam(model.classifier.parameters(), lr=0.001)
```
