# Simple Neural Network

* 1 Input
* 1 Hidden Layer
* 2 Output

```python
import torch
import torch.nn as nn

# Define the neural network
class SimpleNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(SimpleNN, self).__init__()
        self.hidden = nn.Linear(input_size, hidden_size) # Input layer to hidden layer
        self.relu = nn.ReLU() # Activation function
        self.output = nn.Linear(hidden_size, output_size) # Hidden layer to output layer

    def forward(self, x):
        out = self.hidden(x)
        out = self.relu(out)
        out = self.output(out)
        return out
```

```python
# Instantiate the network with 1 input, 1 hidden, and 2 output neurons

input_size = 1
hidden_size = 1
output_size = 2

model = SimpleNN(input_size, hidden_size, output_size)

print(model)
```

```python
# Provide a simple  input
input_data = torch.randn(1, input_size) # Create a random input tensor

# Get the output from the model
output = model(input_data)

# Display the input and output
print("Input:")
display(input_data)
print("\nModel Output:")
display(output)
```

```python
print(f"Input: {input_data.item()}, input length: {input_data.shape}")
print(f"Output: {output.tolist()}, ouput shape {output.shape}")
```
