```python
import torch
```

### 1. Tensor Properties: ndim, shape, size, len
#### 1-D Tensor (Vector)

```python
vector = torch.tensor([1, 2, 3, 4])

print(f"\nVector: {vector}")
print(f"Number of dimensions (ndim): {vector.ndim}") # Number of dimensions
print(f"Shape: {vector.shape}") # Size of each dimension
print(f"Total number of elements (size): {vector.size()}") # Total elements
print(f"Length (len): {len(vector)}") # Length of the first dimension
```

#### 2-D Tensor (Matrix)

```python
matrix = torch.tensor([[1, 2, 3],
                       [4, 5, 6]])
print(f"\nMatrix:\n{matrix}")
print(f"Number of dimensions (ndim): {matrix.ndim}")
print(f"Shape: {matrix.shape}")
print(f"Total number of elements (size): {matrix.size()}")
print(f"Length (len): {len(matrix)}")
```

#### 3-D Tensor

```python
tensor_3d = torch.tensor([[[1, 2],
                           [3, 4]],
                          [[5, 6],
                           [7, 8]]])
                           
print(f"\n3D Tensor:\n{tensor_3d}")
print(f"Number of dimensions (ndim): {tensor_3d.ndim}")
print(f"Shape: {tensor_3d.shape}")
print(f"Total number of elements (size): {tensor_3d.size()}")
print(f"Length (len): {len(tensor_3d)}")
```

## Data Types (`dtypes`)

PyTorch supports various data types for tensors, which determine how the numbers are stored in memory. Common dtypes include `torch.float32` (or `torch.float`), `torch.float64` (or `torch.double`), `torch.int32` (or `torch.int`), `torch.int64` (or `torch.long`), and `torch.bool`.

```python
# Default float dtype is torch.float32
tensor_float = torch.tensor([1.0, 2.0, 3.0])
print(f"\nFloat Tensor: {tensor_float}, Dtype: {tensor_float.dtype}")

# Integer tensor (default is torch.int64 for integers)
tensor_int = torch.tensor([1, 2, 3])
print(f"Integer Tensor: {tensor_int}, Dtype: {tensor_int.dtype}")

# Explicitly define dtype
tensor_float64 = torch.tensor([1.0, 2.0], dtype=torch.float64)
print(f"Float64 Tensor: {tensor_float64}, Dtype: {tensor_float64.dtype}")

tensor_int32 = torch.tensor([10, 20], dtype=torch.int32)
print(f"Int32 Tensor: {tensor_int32}, Dtype: {tensor_int32.dtype}")

# Type casting
tensor_to_cast = torch.tensor([1.5, 2.8, 3.1])
print(f"\nOriginal Tensor: {tensor_to_cast}, Dtype: {tensor_to_cast.dtype}")

tensor_casted_int = tensor_to_cast.long() # Cast to long (int64)
print(f"Casted to Long: {tensor_casted_int}, Dtype: {tensor_casted_int.dtype}")

tensor_casted_float = tensor_int32.float() # Cast int32 to float32
print(f"Casted to Float: {tensor_casted_float}, Dtype: {tensor_casted_float.dtype}")
```

## Element-wise Addition (`torch.add` or `+` operator)

Element-wise addition performs addition on corresponding elements of two tensors. The tensors must have compatible shapes (broadcasting rules apply).

```python
tensor_a = torch.tensor([[1, 2],
                         [3, 4]])
tensor_b = torch.tensor([[5, 6],
                         [7, 8]])

print(f"Tensor A:\n{tensor_a}")
print(f"Tensor B:\n{tensor_b}")

# Element-wise addition using torch.add
add_result_func = torch.add(tensor_a, tensor_b)
print(f"\nAddition using torch.add:\n{add_result_func}")

# Element-wise addition using the + operator
add_result_op = tensor_a + tensor_b
print(f"\nAddition using + operator:\n{add_result_op}")

# Example with broadcasting
tensor_scalar = torch.tensor(10)
broadcast_add = tensor_a + tensor_scalar
print(f"\nTensor A + Scalar (Broadcasting):\n{broadcast_add}")
```

## Concatenation (`torch.cat`)

`torch.cat` is used to concatenate (join) a sequence of tensors along a given dimension. All tensors must either have the same shape (except in the concatenating dimension) or be empty.

```python
tensor1 = torch.tensor([[1, 2],
                        [3, 4]])
tensor2 = torch.tensor([[5, 6],
                        [7, 8]])

print(f"Tensor 1:\n{tensor1}")
print(f"Tensor 2:\n{tensor2}")

# Concatenate along dimension 0 (rows)
cat_dim0 = torch.cat((tensor1, tensor2), dim=0)
print(f"\nConcatenated along dim=0 (rows):\n{cat_dim0}")

# Concatenate along dimension 1 (columns)
cat_dim1 = torch.cat((tensor1, tensor2), dim=1)
print(f"\nConcatenated along dim=1 (columns):\n{cat_dim1}")

# Example with 1D tensors
vec1 = torch.tensor([1, 2, 3])
vec2 = torch.tensor([4, 5, 6])
cat_vec = torch.cat((vec1, vec2), dim=0)
print(f"\nConcatenated 1D tensors: {cat_vec}")
```

## `torch.dot`: Dot Product of 1-D Tensors

`torch.dot` is used to compute the dot product of two 1-D tensors (vectors). The result is a scalar value.

```python
import torch

# Create two 1-D tensors (vectors)
vector_a = torch.tensor([1, 2, 3])
vector_b = torch.tensor([4, 5, 6])

# Compute the dot product using torch.dot
# Result = (1*4) + (2*5) + (3*6) = 4 + 10 + 18 = 32
dot_product = torch.dot(vector_a, vector_b)

print(f"Vector A: {vector_a}")
print(f"Vector B: {vector_b}")
print(f"Dot Product: {dot_product}")
```

## `torch.matmul`: Matrix Multiplication

`torch.matmul` is a more general function for matrix multiplication. It can handle tensors of different dimensions.

### 1. Matrix-Vector Multiplication

```python
# Create a matrix and a vector
matrix = torch.tensor([[1, 2],
                       [3, 4]])
vector = torch.tensor([5, 6])

# Perform matrix-vector multiplication
# Result[0] = (1*5) + (2*6) = 5 + 12 = 17
# Result[1] = (3*5) + (4*6) = 15 + 24 = 39
result_mv = torch.matmul(matrix, vector)

print(f"Matrix: {matrix}")
print(f"Vector: {vector}")
print(f"Matrix-Vector Multiplication Result: {result_mv}")
```

### 2. Matrix-Matrix Multiplication

```python
# Create two matrices
matrix_a = torch.tensor([[1, 2],
                         [3, 4]])
matrix_b = torch.tensor([[5, 6],
                         [7, 8]])

# Perform matrix-matrix multiplication
# Result[0,0] = (1*5) + (2*7) = 5 + 14 = 19
# Result[0,1] = (1*6) + (2*8) = 6 + 16 = 22
# Result[1,0] = (3*5) + (4*7) = 15 + 28 = 43
# Result[1,1] = (3*6) + (4*8) = 18 + 32 = 50
result_mm = torch.matmul(matrix_a, matrix_b)

print(f"Matrix A:{matrix_a}")
print(f"Matrix B:{matrix_b}")
print(f"Matrix-Matrix Multiplication Result:{result_mm}")
```

## `torch.dot` vs `torch.matmul` for 1-D Tensors

When used with two 1-D tensors, `torch.matmul` also computes the dot product. However, there's a subtle difference in the output:
- `torch.dot` returns a **scalar** (a single number).
- `torch.matmul` returns a **0-D tensor** (a tensor containing a single number).

```python
# Using the same vectors from the first example
vector_a = torch.tensor([1, 2, 3])
vector_b = torch.tensor([4, 5, 6])

# Dot product with torch.dot
dot_result = torch.dot(vector_a, vector_b)

# Dot product with torch.matmul
matmul_result = torch.matmul(vector_a, vector_b)

print(f"Result from torch.dot: {dot_result} (Type: {type(dot_result)})")
print(f"Result from torch.matmul: {matmul_result} (Type: {type(matmul_result)})")
```

## Transposing a Matrix (`.T` or `transpose()`)

Transposing a matrix means flipping it over its diagonal, i.e., rows become columns and columns become rows. In PyTorch, you can use the `.T` attribute or the `transpose()` method.

```python
# Original matrix
matrix_orig = torch.tensor([[1, 2, 3],
                            [4, 5, 6]])
print(f"Original Matrix:\n{matrix_orig}")
print(f"Shape of Original Matrix: {matrix_orig.shape}")

# Transpose using .T
matrix_transposed_T = matrix_orig.T
print(f"\nTransposed Matrix (.T):\n{matrix_transposed_T}")
print(f"Shape of Transposed Matrix (.T): {matrix_transposed_T.shape}")

# Transpose using transpose(dim0, dim1)
# For a 2D tensor, dim0=0 (rows), dim1=1 (columns)
matrix_transposed_func = torch.transpose(matrix_orig, 0, 1)
print(f"\nTransposed Matrix (transpose(0,1)):\n{matrix_transposed_func}")
print(f"Shape of Transposed Matrix (transpose(0,1)): {matrix_transposed_func.shape}")
```

## `state_dict` and `parameters` (for Neural Networks)

When working with neural networks in PyTorch, `state_dict` and `parameters` are crucial for saving, loading, and inspecting the model's learnable weights and biases.

- `state_dict()`: Returns a dictionary containing a whole state of the module. This includes all parameters and persistent buffers.
- `parameters()`: Returns an iterator over module parameters. These are the tensors that are updated during training.

```python
# Creating a simple neural network
class SimpleNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear1 = nn.Linear(10, 5) # 10 input features, 5 output features
        self.relu = nn.ReLU()
        self.linear2 = nn.Linear(5, 1)  # 5 input features, 1 output feature

    def forward(self, x):
        x = self.linear1(x)
        x = self.relu(x)
        x = self.linear2(x)
        return x

model = SimpleNN()

print("\nModel's state_dict:")
for param_tensor, value in model.state_dict().items():
    print(f"{param_tensor}\t{value.size()}")

print("\nModel's learnable parameters:")
for name, param in model.named_parameters():
    if param.requires_grad:
        print(f"{name}\t{param.size()}")
```

```python
# Display available variables and their types/values
print("Available variables:")
vars_to_show = {
    'model': model,
    'name': name,
    'param': param,
    'param_tensor': param_tensor,
    'value': value
}
for k, v in vars_to_show.items():
    try:
        print(f"{k}: type={type(v)}, value={v}")
    except Exception:
        print(f"{k}: type={type(v)}, value=<<unprintable>>")

# Show model state_dict keys and shapes
print("\nModel's state_dict keys and shapes:")
for k, v in model.state_dict().items():
    print(f"{k}\t{v.size()}")

# Print model's learnable parameters (as requested)
print("\nModel's learnable parameters:")
for name, param in model.named_parameters():
    if param.requires_grad:
        print(f"{name}\t{param.size()}")
```
