# Single Neuron with PyTorch

## 1. Import PyTorch
Import the PyTorch library.

```python
import torch
```

## 2. Define Tensors for Inputs, Weights, and Bias
Define the input features, weights, and bias for the neuron as PyTorch tensors.

```python
x = torch.tensor([0.6, 0.4])
w = torch.tensor([0.4, -0.2]) # try with 'dtype' and 'requires_grad'
b = torch.tensor(0.5) # try with 'dtype' and 'requires_grad'
```

## 3. Calculate the Linear Combination
Calculate the weighted sum of the inputs and add the bias.

```python
z = torch.dot(x, w) + b  # Linear
```

## 4. Output Before Activation
Display the output of the linear operation before applying any activation function.

```python
f'Output: No Activation {z}'
```

## 5. Apply Activation Functions
Apply different activation functions (ReLU, Sigmoid, Tanh) to the linear output using PyTorch functions.

```python
y_relu = torch.relu(z)
y_sigmoid = torch.sigmoid(z)
y_tanh = torch.tanh(z)
```

## 6. Outputs After Activation
Display the outputs of the neuron after applying the activation functions.

```python
print(f'Relu: Weighted sum: {z} -- Output after Activation function: {y_relu}')
print(f'Sigmoid: Weighted sum: {z} -- Output after Activation function: {y_sigmoid}')
print(f'Tanh: Weighted sum: {z} -- Output after Activation function: {y_tanh}')
```

## 7. Classification based on Sigmoid Output
Use the sigmoid output to perform a binary classification.

```python
cls = torch.where(y_sigmoid >= 0.5, torch.tensor(1), torch.tensor(0))
cls
```

## 8. Backpropagation and Gradient Descent

**Backpropagation** is the core algorithm for training neural networks. It works by calculating the gradient of the loss function with respect to each parameter (weight and bias) in the network. These gradients represent the direction and magnitude of the change needed to reduce the loss.

The process involves two main passes:
1.  **Forward Pass:** Input data is fed through the network to produce an output, and the loss (the error between the predicted output and the true target) is calculated.
2.  **Backward Pass:** The gradients of the loss are calculated with respect to the network's parameters, starting from the output layer and moving backward to the input layer. This is done using the chain rule of calculus.

**Gradient Descent** is an optimization algorithm that uses these gradients to update the network's parameters. The parameters are adjusted in the opposite direction of the gradient to minimize the loss. The size of the update step is controlled by the **learning rate**.

The update rule for a parameter `p` is: `p = p - learning_rate * gradient_of_p`

In the following cell, we will:
1.  Perform a forward pass to calculate the loss.
2.  Perform a backward pass (backpropagation) to compute gradients.
3.  Update the weights and bias using gradient descent.
4.  Perform another forward pass to show that the loss has decreased.

#### --- Step 1: Initialization ---

```python
# Input and true target
x = torch.tensor([0.6, 0.4])
y_true = torch.tensor(1.0)

# Initial weights and bias with gradient tracking enabled
w = torch.tensor([0.4, -0.2], requires_grad=True)
b = torch.tensor(0.5, requires_grad=True)

# Learning rate
lr = 0.1
```

```python
print(f'Input: {x}')
print(f'Input shape: {x.shape},Input ndim: {x.ndim}, Input dtype: {x.dtype}')
print(f'Input size: {x.size()}, Input stride: {x.stride()}')
print(f'Input data: {x.data}')
print(f'Input device: {x.device}, Input requires_grad: {x.requires_grad}, Input numel: {x.numel()}')
print('-' * 20)
print(f'y_true: {y_true}, y_true data: {y_true.data}, y_true dtype: {y_true.dtype}, y_true device: {y_true.device}, y_true requires_grad: {y_true.requires_grad}')
print('-' * 20)
print(f'Initial weights: {w.data}')
print(f'Initial bias: {b.data}')
print('-' * 20)
```

### --- Step 2: Forward Pass --- Calculate the predicted output and the loss

```python
z = torch.dot(x, w) + b
y_pred = torch.sigmoid(z)
loss = (y_pred - y_true)**2  # Mean Squared Error loss

print(f'Predicted output: {y_pred.item()}')
print(f'Initial loss: {loss.item()}')
print('-' * 20)
```

### --- Step 3: Backward Pass (Backpropagation) --- Compute gradients of the loss with respect to w and b

```python
loss.backward()

print(f'Weight gradients: {w.grad}')
print(f'Bias gradient: {b.grad}')
print('-' * 20)
```

### --- Step 4: Update Parameters (Gradient Descent) --- Update weights and bias using the gradients
#### We use torch.no_grad() to prevent this operation from being part of the computation graph

```python
with torch.no_grad():
    w -= lr * w.grad
    b -= lr * b.grad

print(f'Updated weights: {w.data}')
print(f'Updated bias: {b.data}')
print('-' * 20)
```

### --- Step 5: Zero Gradients ---  Zero out the gradients for the next iteration

```python
w.grad.zero_()
b.grad.zero_()
```

### --- Step 6: Verify the update --- Perform another forward pass to see if the loss has decreased

```python
z_new = torch.dot(x, w) + b
y_pred_new = torch.sigmoid(z_new)
loss_new = (y_pred_new - y_true)**2

print(f'New predicted output: {y_pred_new.item()}')
print(f'New loss: {loss_new.item()}')
```
