```python
import torch
```

```python
x = torch.tensor([0.6, 0.4])
w = torch.tensor([0.4, -0.2]) # try with 'dtype' and 'requires_grad'
b = torch.tensor(0.5) # try with 'dtype' and 'requires_grad'
```

```python
z = torch.dot(x, w) + b  # Linear
```

```python
f"Output: No Activation {z}"
```

```python
y_relu = torch.relu(z)
y_sigmoid = torch.sigmoid(z)
y_tanh = torch.tanh(z)
```

```python
print(f"Relu: Weighted sum: {z} -- Output after Activation function: {y_relu}")
print(f"Sigmoid: Weighted sum: {z} -- Output after Activation function: {y_sigmoid}")
print(f"Tanh: Weighted sum: {z} -- Output after Activation function: {y_tanh}")
```

```python
cls = torch.where(y_sigmoid >= 0.5, torch.tensor(1), torch.tensor(0))
cls
```

```python
import matplotlib.pyplot as plt
```

```python
z1 = torch.linspace(-5, 5, 100)
```

```python
plt.figure(figsize=(8,5))
plt.plot(z1.numpy(), torch.sigmoid(z1).numpy(), label='Sigmoid', color='blue')
plt.plot(z1.numpy(), torch.relu(z1).numpy(), label='ReLU', color='green')
plt.plot(z1.numpy(), torch.tanh(z1).numpy(), label='Tanh', color='red')

plt.title("Activation Functions")
plt.xlabel("z1 - linear space")
plt.ylabel("Activation output")
plt.legend()
plt.grid(True)
plt.show()
```
