# XOR Problem

The XOR problem is a classic example in deep learning which demonstrates that a single-layer perceptron cannot solve nonlinear problems like XOR. To solve XOR, a multi-layer perceptron (MLP) with at least one hidden layer and nonlinear activation functions is needed.

**Problem Overview**:
- XOR (exclusive OR) outputs 1 when the inputs differ, and 0 otherwise.

- It is not linearly separable, meaning a single-layer perceptron cannot learn XOR.

- Neural networks overcome this by transforming input space using hidden layers and nonlinear activation functions to create linearly separable features.

## XOR with NumPy
**Overview**:
Build a small neural network manually with:
- An input layer (2 neurons),
- A hidden layer (at least 2 neurons) with nonlinear activation (e.g., sigmoid or ReLU),
- An output layer (1 neuron) with sigmoid activation.

Train using gradient descent and backpropagation

```python
import numpy as np
```

```python
# XOR inputs and expected outputs
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])
y = np.array([[0], [1], [1], [0]]
```

```python
# Sigmoid activation and its derivative
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)
```

```python
np.random.seed(42)
```

```python
# Initialize weights and biases, layesr
input_layer_neurons = 2
hidden_layer_neurons = 2
output_neurons = 1

wh = np.random.uniform(size=(input_layer_neurons, hidden_layer_neurons))
bh = np.random.uniform(size=(1, hidden_layer_neurons))
wout = np.random.uniform(size=(hidden_layer_neurons, output_neurons))
bout = np.random.uniform(size=(1, output_neurons))
```

```python
# Training parameters
lr = 0.1
epochs = 1000
```

```python
# Training loop
for epoch in range(epochs):
    # Forward propagation
    hidden_layer_input = np.dot(X, wh) + bh
    hidden_layer_activations = sigmoid(hidden_layer_input)
    output_layer_input = np.dot(hidden_layer_activations, wout) + bout
    predicted_output = sigmoid(output_layer_input)

    # Backpropagation
    error = y - predicted_output
    d_predicted_output = error * sigmoid_derivative(predicted_output)

    error_hidden_layer = d_predicted_output.dot(wout.T)
    d_hidden_layer = error_hidden_layer * sigmoid_derivative(hidden_layer_activations)

    # Update weights and biases
    wout += hidden_layer_activations.T.dot(d_predicted_output) * lr
    bout += np.sum(d_predicted_output, axis=0, keepdims=True) * lr
    wh += X.T.dot(d_hidden_layer) * lr
    bh += np.sum(d_hidden_layer, axis=0, keepdims=True) * lr
```

```python
# Output predictions
print("Final predicted output:")
print(np.round(predicted_output))
```

## XOR with Torch (PyTorch)
**Overview**: Define a MLP using torch.nn
- Using Linear layers for hidden and ouput.
- Using nonlinear activations like ReLU or Sigmoid.
- Use SGD or Adam as an optimizer.
- Use BCELoss to calculate loss function.

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# Data
X = torch.tensor([[0,0],[0,1],[1,0],[1,1]], dtype=torch.float32)
y = torch.tensor([[0],[1],[1],[0]], dtype=torch.float32)
```

```python
# Define the network
class XORNetwork(nn.Module):
    def __init__(self):
        super(XORNet, self).__init__()
        self.hidden = nn.Linear(2, 2)  # Input to hidden
        self.output = nn.Linear(2, 1)  # Hidden to output
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.sigmoid(self.hidden(x))
        x = self.sigmoid(self.output(x))
        return x
```

```python
# Create the model, loss function and optimizer
model = XORNetwork()
criterion = nn.BCELoss()
optimizer = optim.SGD(model.parameters(), lr=0.1) # learning rate
```

```python
# Training loop
epochs = 10000
for epoch in range(epochs):
    optimizer.zero_grad()
    output = model(X)
    loss = criterion(output, y)
    loss.backward()
    optimizer.step()
```

```python
# Predictions
with torch.no_grad():
    predictions = model(X)
    print("Predicted outputs:")
    print(torch.round(predictions))
```
